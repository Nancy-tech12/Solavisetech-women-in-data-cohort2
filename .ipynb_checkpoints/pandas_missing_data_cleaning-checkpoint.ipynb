```python
import pandas as pd
import numpy as np
```

```python
data = {
'Order_ID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
'Customer_Name': ['Amara', 'Bola', None, 'Diana', 'Efua', 'Fatima', 'Grace',
None, 'Ifeoma', 'Joyce'],
'Country': ['Ghana', 'Nigeria', 'Kenya', None, 'Ghana', 'Cameroon', None,
'Kenya', 'Nigeria', 'Ghana'],
'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', None, 'Phone', 'Tablet',
'Laptop', None, 'Phone'],
'Quantity': [2, 1, None, 3, 2, None, 1, 4, 2, None],
'Price': [1500, 800, 600, 1500, None, 800, 600, 1500, None, 800],
'Delivery_Days': [3, None, 7, 5, 3, None, 6, None, 4, 5]
}
df = pd.DataFrame(data)
print(df)
```

```python
df.info()
df.describe()
df.isnull().sum()
```

The dataset contains 10 orders across 7 columns. Upon inspection, Quantity and Delivery_Days have the most missing values with 3 each, while Customer_Name, Country, Product and Price each have 2 missing values. Order_ID is the only column with no missing values, confirming it can serve as a reliable unique identifier for each order.

```python
df["Customer_Name"] = df["Customer_Name"].fillna("Unknown")
```

Missing customer names were filled with "Unknown" rather than dropping the rows entirely, because the remaining order details such as product, price and quantity are still valuable for analysis and should not be lost.

```python
df["Price"] = df["Price"].fillna(df["Price"].mean())
df["Quantity"] =df["Quantity"].fillna(df["Quantity"].mean())
print(df)
```

The mean was used to fill missing Price values because the prices are relatively consistent with no extreme outliers, making the average a reliable estimate for the missing figures. The mean was also used to fill missing Quantity values because the quantities range from 1 to 4 with no extreme outliers, making the average a fair estimate for the missing values.

```python
df["Delivery_Days"].describe()
```

```python
df["Delivery_Days"] = df["Delivery_Days"].fillna(df["Delivery_Days"].mean())
```

After checking for outliers using describe(), the Delivery_Days values range from 3 to 7 with no extreme values, therefore the mean was used to fill the missing values as it accurately represents the typical delivery time.

```python
df["Country"] = df["Country"].fillna("Unknown")
print(df)
```

Missing country values were filled with "Unknown" rather than dropping the rows, because the remaining order details such as product, price and quantity are still valuable for analysis. Even an "Unknown" country flags a data collection issue worth investigating later.

```python
df = df.dropna(subset=["Product"])
print(df)
```

Rows with missing product values were dropped entirely because without knowing what was sold, the order has no analytical value. Revenue cannot be calculated, inventory cannot be tracked and sales patterns cannot be analysed.

REFLECTION

The hardest task for me was Task 5, where I had to decide what to do with the Country and Product columns. These are not numbers so I could not just calculate a mean or median to fill them in. I had to stop and think about what made business sense. For Country, I filled the missing values with Unknown because the rest of the order information was still useful. But for Product, I decided to drop those rows completely because if you do not know what was sold, the order tells you nothing useful at all. No, I did not use the same method for every column. Customer Name and Country got Unknown, Price and Quantity were filled with the mean because their values were close together with no extreme numbers, and Delivery Days was also filled with the mean after I checked and confirmed there were no outliers. Every decision was based on what made the most sense for that specific column. If 50% of the Price column was missing instead of just a few values, I would not fill it with the mean. That would mean half the prices in the dataset are guesses and not real numbers. Any analysis done on that data would not be trustworthy. In that case I would either investigate why so many prices were not recorded or consider dropping the Price column altogether.
