```python
import pandas as pd
import numpy as np
```

```python
retail_data = {
    "Product":  ["Shoes", "Bag", "Shoes", None, "Watch"],
    "Category": ["Footwear", "Accessories", "Footwear", "Clothing", None],
    "Price":    [5000, 8000, 5000, None, 25000],
    "Quantity": [10, None, 10, 5, 3]
}

df = pd.DataFrame(retail_data)
print(df)
```

```python
df.info()
df.isnull().sum()
```

```python
df = df.drop_duplicates()
print(df)
```

```python
df["Price"] = df["Price"].fillna(df["Price"].mean())
print(df)
```

```python
df = df.dropna(subset={"Product"})
print(df)
```

```python
df = df.rename(columns={"Quantity":"Units Sold"})
print(df)
```

```python
df.isnull().sum()
```

```python
df["Category"] = df["Category"].fillna("Unknown")
df["Units Sold"] = df["Units Sold"].fillna(df["Units Sold"].mean())
print(df)
```

```python
df.isnull().sum()
```

```python
df = df.to_csv("cleaned_retail_data.csv", index=False)
print("Saved successfully!")
```
