```python
import pandas as pd
```

```python
data = {
    "Region":   ["North","South","East","West","North","East","South","West"],
    "Category": ["Electronics","Clothing","Electronics","Clothing",
                 "Food","Food","Electronics","Food"],
    "Month":    ["Jan","Jan","Feb","Feb","Mar","Mar","Apr","Apr"],
    "Sales":    [1200, 800, 1500, 600, 900, 1100, 1300, 750],
    "Units":    [12, 20, 15, 18, 25, 22, 13, 30]
}

df = pd.DataFrame(data)
print(df)
```

Five Business Questions
01.	Which region generated the highest total revenue?
02.	Which product category had the most units sold?
03.	Which month had the highest total sales?
04.	What is the average sale value for the Electronics category?
05.	Which region had the lowest average sales?

```python
region_sales = df.groupby("Region")["Sales"].sum()
print(region_sales)
```
